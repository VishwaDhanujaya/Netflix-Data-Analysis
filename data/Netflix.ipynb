## Dataset Overview

The dataset contains information about movies and TV shows available on Netflix, including:

- Title
- Type (Movie or TV Show)
- Director and Cast
- Country of production
- Date added to Netflix
- Release year
- Rating
- Duration
- Genres

Source: Kaggle Netflix Dataset

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("netflix_titles.csv")

df.head()
df.info()
df.describe()
```

## Data Cleaning

Before analysis, the dataset was cleaned to ensure accuracy and consistency.

Steps performed:
- Identified missing values
- Filled or handled null entries
- Converted date columns to datetime format
- Created new features for analysis

```python
df.fillna({
    'country': "Unknown",
    'director': "Unknown",
    'cast': "Unknown"
}, inplace=True)
```

```python
df['date_added'] = df['date_added'].str.strip()
df['date_added'] = pd.to_datetime(df['date_added'], errors='coerce')
df['year_added'] = df['date_added'].dt.year
```

## Exploratory Data Analysis

Exploratory Data Analysis was performed to understand patterns and relationships within the dataset.

The analysis investigates:
- Distribution of content types
- Yearly content additions
- Country contributions
- Genre distribution

### Movies vs TV Shows Distribution

```python
sns.countplot(x='type', data=df)
plt.title("Movies vs TV Shows")
plt.show()
```

### Content Added Over Years

```python
df['year_added'].value_counts().sort_index().plot(kind='line')
plt.title("Content Added Per Year")
plt.show()
```

### Top Content Producing Countries

```python
df['country'].value_counts().head(10).plot(kind='bar')
plt.title("Top 10 Content Producing Countries")
plt.show()
```

### Most Popular Genres

```python
genres = df['listed_in'].str.split(', ')
genre_list = sum(genres, [])
genre_counts = pd.Series(genre_list).value_counts().head(10)

genre_counts.plot(kind='bar')
plt.title("Top Genres")
plt.show()
```

## Key Insights

- Netflix hosts significantly more Movies than TV Shows.
- Content additions increased rapidly after 2015.
- The United States contributes the largest portion of content.
- Drama and International Movies are among the most popular genres.
- Netflix expanded aggressively during the late 2010s.

These insights demonstrate how Netflix's content strategy evolved over time.
